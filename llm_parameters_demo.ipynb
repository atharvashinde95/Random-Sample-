## 🎛️ What are these Parameters?

| Parameter   | What it controls              | Range  | Low = ?             | High = ?            |
|-------------|-------------------------------|--------|---------------------|---------------------|
| Temperature | Randomness of output          | 0 – 2  | Focused, repetitive | Creative, random    |
| Top P       | Vocabulary pool by probability| 0 – 1  | Narrow word choice  | Wider word choice   |
| Top K       | Top K words to sample from    | 1 – N  | Very restricted     | More variety        |

> 💡 **Rule of thumb:**
> - Factual tasks (Q&A, code) → Low Temperature `0.0 – 0.3`
> - Creative tasks (stories, ideas) → High Temperature `0.7 – 1.2`
> - Use **either** Top P or Top K, not both together


## 🌡️ Temperature Demo

```
prompt = "Describe the sun in one sentence."

print("=" * 60)
print(f"📝 Prompt: {prompt}")
print("=" * 60)

for temp in [0.0, 0.7, 1.5]:
    llm_temp = ChatOpenAI(
        model=os.getenv("MODEL"),
        base_url=os.getenv("API_URL"),
        api_key=os.getenv("API_KEY"),
        temperature=temp,
    )
    response = llm_temp.invoke(prompt)
    print(f"\n🌡️  Temperature = {temp}")
    print(f"   {response.content}")
```

## 🎯 Top P Demo

```
prompt = "Give me a creative name for a coffee shop."

print("=" * 60)
print(f"📝 Prompt: {prompt}")
print("=" * 60)

for top_p in [0.1, 0.5, 1.0]:
    llm_topp = ChatOpenAI(
        model=os.getenv("MODEL"),
        base_url=os.getenv("API_URL"),
        api_key=os.getenv("API_KEY"),
        temperature=1.0,
        top_p=top_p,
    )
    response = llm_topp.invoke(prompt)
    print(f"\n🎯 Top P = {top_p}")
    print(f"   {response.content}")
```

## ⚖️ Temperature vs Top P — Side by Side Comparison

```
prompt = "Write an opening line for a mystery novel."

configs = [
    {"label": "Precise  (low temp, low top_p)",  "temperature": 0.1, "top_p": 0.1},
    {"label": "Balanced (mid temp, mid top_p)",  "temperature": 0.7, "top_p": 0.5},
    {"label": "Creative (high temp, high top_p)","temperature": 1.2, "top_p": 0.95},
]

print("=" * 60)
print(f"📝 Prompt: {prompt}")
print("=" * 60)

for cfg in configs:
    llm_cfg = ChatOpenAI(
        model=os.getenv("MODEL"),
        base_url=os.getenv("API_URL"),
        api_key=os.getenv("API_KEY"),
        temperature=cfg["temperature"],
        top_p=cfg["top_p"],
    )
    response = llm_cfg.invoke(prompt)
    print(f"\n⚙️  {cfg['label']}")
    print(f"   Temp={cfg['temperature']} | Top_P={cfg['top_p']}")
    print(f"   → {response.content}")
```

## 📋 Practical Recommendations by Use Case

```
use_cases = {
    "✅ Code Generation / Bug Fix":     {"temperature": 0.0, "top_p": 0.1},
    "✅ Factual Q&A / Summarization":   {"temperature": 0.2, "top_p": 0.3},
    "✅ Chatbot / General Assistant":   {"temperature": 0.7, "top_p": 0.9},
    "✅ Creative Writing / Brainstorm": {"temperature": 1.0, "top_p": 0.95},
    "✅ Poetry / Story Generation":     {"temperature": 1.2, "top_p": 1.0},
}

print(f"{'Use Case':<40} {'Temperature':>12} {'Top P':>8}")
print("-" * 62)
for use_case, params in use_cases.items():
    print(f"{use_case:<40} {params['temperature']:>12} {params['top_p']:>8}")
```

## 🔢 Top K — Conceptual Demo

```
# ⚠️ NOTE: ChatOpenAI / OpenAI-compatible APIs do NOT support top_k directly.
# Top K is natively supported in Google (Gemini), Anthropic (Claude), Ollama etc.
# For your Capgemini setup, stick to temperature + top_p.

vocab_example = ["happy", "joyful", "glad", "content", "pleased", "ecstatic", "elated", "blissful"]

print("📖 Conceptual: Top K Sampling\n")
print(f"Full vocabulary pool (simplified): {vocab_example}\n")

for k in [2, 4, 8]:
    print(f"Top K = {k}  →  Model only picks from: {vocab_example[:k]}")
```

## ✅ Summary

| Parameter   | Supported in ChatOpenAI? | Best for                        |
|-------------|--------------------------|----------------------------------|
| Temperature | ✅ Yes                   | Primary randomness control       |
| Top P       | ✅ Yes                   | Secondary vocabulary control     |
| Top K       | ❌ No (OpenAI-style API) | Use with Gemini / Claude / Ollama|

### 🔑 Key Takeaways
- **Temperature = 0** → deterministic, same answer every time
- **Temperature > 1** → very creative, may hallucinate more
- **Top P = 0.1** → model picks from only the top 10% probable words
- **Top P = 1.0** → model can pick from the entire vocabulary
- Always set `.env` values and load with `python-dotenv` — never hardcode keys!
